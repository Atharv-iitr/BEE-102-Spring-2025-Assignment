```python
import gzip
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math
from collections import defaultdict
def build_markov_transition_matrix(sequence, k=1):
    counts = defaultdict(lambda: defaultdict(int))
    nucleotides = ['A', 'C', 'G', 'T']

    for i in range(len(sequence) - k):
        current_kmer = sequence[i:i+k]
        next_nuc = sequence[i+k]
        counts[current_kmer][next_nuc] += 1
    transition_matrix = np.zeros((4**k, 4))
    if k == 1:
        transition_matrix = np.zeros((4, 4))
        nuc_to_idx = {'A': 0, 'C': 1, 'G': 2, 'T': 3}
        
        for current in nucleotides:
            total = sum(counts[current].values())
            if total > 0:
                for next_nuc in nucleotides:
                    transition_matrix[nuc_to_idx[current], nuc_to_idx[next_nuc]] = \
                        counts[current][next_nuc] / total
    return transition_matrix


dna_sequence = "ACGTTGCAAGTCAGAATGCATAGACACAGATACGATACAGAT" 
transition_matrix = build_markov_transition_matrix(dna_sequence)
nucleotides = ['A', 'C', 'G', 'T']
print("\t" + "\t".join(nucleotides))
for i, nuc in enumerate(nucleotides):
    print(nuc + "\t" + "\t".join(f"{p:.3f}" for p in transition_matrix[i]))
```
