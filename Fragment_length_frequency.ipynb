```python

import gzip
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math

fragment_lengths = []
with gzip.open('query.bed.gz', 'rt') as file:
    for line in file:
        columns = line.strip().split('\t')
        start = int(columns[1])
        end = int(columns[2])
        length = end - start
        fragment_lengths.append(length)
length_counts = pd.Series(fragment_lengths).value_counts().sort_index()
normalized_counts = length_counts / length_counts.sum()
plt.figure(figsize=(10,6))
plt.plot(normalized_counts.index, normalized_counts.values, linestyle='-')  
plt.xlabel('Fragment Length')
plt.ylabel('Normalized Frequency [A.U.]')
plt.title('Fragment Length Distribution')
plt.grid(True)
plt.show()
```
