```python
import gzip
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math
def convert_fasta_to_single_line(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        current_header = None
        sequence_buffer = []
        
        for line in infile:
            line = line.strip()
            if line.startswith('>'):
                if current_header is not None:
                    outfile.write(''.join(sequence_buffer) + '\n')
                outfile.write(line + '\n')
                current_header = line
                sequence_buffer = []
            else:
                sequence_buffer.append(line)
        if current_header is not None and sequence_buffer:
            outfile.write(''.join(sequence_buffer) + '\n')

if __name__ == "__main__":
    input_file = 'multiline_input.fasta'
    output_file = 'single_line_output.fasta'
    convert_fasta_to_single_line(input_file, output_file)
```
